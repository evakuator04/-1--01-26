```python
import math
def f(x):
  return math.sqrt(1-x)-math.tan(x)
L=0
R=1
eps=0.0001
step=0
while abs(R-L)>eps:
  step+=1
  B=(L+R)/2
  print(step,'.',round(B,4))
  if f(L)*f(B)<=0:
    R=B
  else:
    L=B
answer=(L+R)/2
print(round(answer,4)) #для 13 варианта

```
